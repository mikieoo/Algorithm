```python
import numpy as np

data = [10,20,30,40,50]

mean = np.mean(data)
std = np.std(data)

print('원본 데이터: ',data)

print('[표준화 결과]')
standard = (data-mean)/std
print(standard)

print(['정규화 결과'])
normalization = data - np.min(data) / np.max(data) - np.min(data)
print(normalization)
```
